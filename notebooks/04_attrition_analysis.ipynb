```python
from scipy.stats import chi2_contingency
contingency=pd.crosstab(hr.OverTime,hr.Attrition)
chi2,p,dof,expected=chi2_contingency(contingency)
print({'chi_square':chi2,'p':p,'cramers_v':(chi2/len(hr))**.5})
display(contingency)
print('Exploratory fictional association; no causal workload or retention estimate.')
```
